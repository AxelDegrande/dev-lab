```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
def perlinNoise(length):
    data = np.array([np.random.normal(loc=0, scale=1)])


    for i in range(length):
        data = np.append(data, np.random.normal(loc=data[i], scale=0.1))

    return data
        
```

```python
plt.plot(np.arange(0, 1001), perlinNoise(1000))
```
