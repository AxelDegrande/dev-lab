Use a custom probability distribution to vary the size of the random walker’s steps. The step size can be determined by influencing the range of values picked with a qualifying random value. Can you map the probability to a quadratic function by making the likelihood that a value is picked equal to the value squared?

```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
class Walker:
    def __init__(self, startX, startY):
        self.x = startX
        self.y = startY
        self.xyArray = np.array([[self.x, self.y]])

    def step(self):
        stepChance = np.random.rand()

        if stepChance < 0.01:
            stepX = np.random.choice([-25, 25], p=[0.5, 0.5])
            stepY = np.random.choice([-25, 25], p=[0.5, 0.5])
        else:
            stepX = np.random.randint(-1, 2)
            stepY = np.random.randint(-1, 2)
        
        self.x += stepX
        self.y += stepY

    def update(self):
        self.step()
        self.xyArray = np.vstack([self.xyArray, [self.x, self.y]])
        
```

```python
walker = Walker(0, 0)

for _ in range(1000):
    walker.update()
```

```python
plt.figure(figsize=(8, 8))
line, = plt.plot(walker.xyArray[:, 0], walker.xyArray[:, 1])
plt.scatter(walker.xyArray[0, 0], walker.xyArray[0, 1], color='green', label='Start')
plt.scatter(walker.xyArray[-1, 0], walker.xyArray[-1, 1], color='red', label='End')
plt.legend()
plt.axis('equal') 
plt.show()
```
