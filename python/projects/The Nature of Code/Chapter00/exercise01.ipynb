## Create a random walker that has a greater tendency to move down and to the right.

```python
import matplotlib.pyplot as plt
import numpy as np 
```

```python
class Walker:
    def __init__(self, startX, startY):
        self.x = startX
        self.y = startY
        self.xyArray = np.array([[self.x, self.y]])

    def step(self):
        stepX = np.random.choice([-1, 0, 1], p=[0.3, 0.3, 0.4])
        stepY = np.random.choice([-1, 0, 1], p=[0.4, 0.3, 0.3])
        
        self.x += stepX
        self.y += stepY

    def update(self):
        self.step()
        self.xyArray = np.vstack([self.xyArray, [self.x, self.y]])
        
```

```python
walker = Walker(0, 0)

for _ in range(1000):
    walker.update()
```

```python
plt.figure(figsize=(8, 8))
line, = plt.plot(walker.xyArray[:, 0], walker.xyArray[:, 1])
plt.scatter(walker.xyArray[0, 0], walker.xyArray[0, 1], color='green', label='Start')
plt.scatter(walker.xyArray[-1, 0], walker.xyArray[-1, 1], color='red', label='End')
plt.legend()
plt.axis('equal') 
plt.show()
```
