```python
import pygame
import math
```

```python
class Ball:
    def __init__(self, screen, SCREEN_WIDTH, SCREEN_HEIGHT, BALL_RADIUS, BALL_POS_X):
        self.screen = screen
        self.screenWidth = SCREEN_WIDTH
        self.screenHeight = SCREEN_HEIGHT
        self.ballRadius = BALL_RADIUS
        self.position = pygame.Vector2(BALL_POS_X, self.ballRadius + 10) # Random start position in the X range, Y starts at bottom of screen
        self.velocity = pygame.Vector2(0, 0)
        self.acceleration = pygame.Vector2(0, 0)
        self.netForce = pygame.Vector2(0, 0)
        self.mass = 0.1*math.pi*self.ballRadius**2 # Mass of the object is defined by the area of the cicle
        self.maxVelocity = 10
        self.maxAcceleration = 0.1
        
    def draw(self):
        pygame.draw.circle(self.screen, color='red', center=self.position, radius=self.ballRadius)

    def move(self):
        self.acceleration = self.netForce / self.mass
        self.accelerationLimit()
        self.velocity += self.acceleration
        self.velocityLimit()
        self.position += self.velocity

    def velocityLimit(self):
        self.velocity.x = max(-self.maxVelocity, min(self.velocity.x, self.maxVelocity))
        self.velocity.y = max(-self.maxVelocity, min(self.velocity.y, self.maxVelocity))

    def accelerationLimit(self):
         self.acceleration.x = max(-self.maxAcceleration, min(self.acceleration.x, self.maxAcceleration))
         self.acceleration.y = max(-self.maxAcceleration, min(self.acceleration.y, self.maxAcceleration))

    def collisionDetection(self):
        if self.position.x > (self.screenWidth-self.ballRadius):
            self.position.x = self.screenWidth - self.ballRadius
            self.velocity.x *= -0.9

        if self.position.x < (0+self.ballRadius):
            self.position.x = self.ballRadius
            self.velocity.x *= -0.9

        if self.position.y > (self.screenHeight-self.ballRadius):
            self.position.y = self.screenHeight - self.ballRadius
            self.velocity.y *= -0.9

        if self.position.y < (0+self.ballRadius):
            self.position.y = self.ballRadius
            self.velocity.y *= -0.9

   
    def applyForce(self):
        # 1) Collect every force acting on the ball THIS frame.
        #    The list is rebuilt each frame, so forces that are not
        #    active simply never get added.
        self.forces = []

        # Gravity
        self.forces.append(pygame.Vector2(0, 1) * self.mass)

        # Force when key is pressed
        keys = pygame.key.get_pressed()
        if keys[pygame.K_a]:
            self.forces.append(pygame.Vector2(-0.5, 0))
        if keys[pygame.K_d]:
            self.forces.append(pygame.Vector2(0.5, 0))    

        # 2) Sum forces -> net force
        for f in self.forces:
            self.netForce += f


    def update(self):
        self.draw()
        self.applyForce()
        self.move()
        self.collisionDetection()
```

```python

#====CONSTANTS====
SCREEN_WIDTH = 800
SCREEN_HEIGHT = 400
BALL_RADIUS = 24


#====VARIABLES
mouseClick = 0

#====MAIN====
pygame.init()
screen = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))
pygame.display.set_caption("Forces demo: arrows = push")
font = pygame.font.SysFont(None, 24)
clock = pygame.time.Clock()

gameRunning = True

ball_1 = Ball(screen, SCREEN_WIDTH, SCREEN_HEIGHT, BALL_RADIUS, 100)
#ball_2 = Ball(screen, SCREEN_WIDTH, SCREEN_HEIGHT, BALL_RADIUS*2, 200)

while gameRunning:
    # Poll for events
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            gameRunning = False

    # Fill screen with color to wipe away anything from last frame
    screen.fill("white")

    #====MAIN CONTENT====
    ball_1.update()
    #ball_2.update()
    ball_1.draw.o


    #====RENDER TEXT====
    info = f"accel: ({ball_1.velocity.x:.0f}, {ball_1.velocity.y:.0f})"
    screen.blit(font.render(info, True, (250, 230, 250)), (10, 10))

    # Flip the display to put you work on screen
    pygame.display.flip()
    # Limit FPS to 60
    clock.tick(60)
    
pygame.quit()    
```
